Question 1
---
A random sample has
   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$
   Compute the estimated standard error of $\bar{X}$. Test
   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$
   at significance levels $\alpha=.10,.05,.01$ using the large-sample $z$-statistic. These correspond to 90%, 95%, and 99% confidence levels. Construct a 95% confidence interval for $\mu$.


```python
import numpy as np
import pandas as pd
from scipy.stats import norm
 
```

```python
x_bar = 52
Ho = 50
s = 10
n = 100


std_err = s/ np.sqrt(n)
print('standard error: ', std_err)

# test Ho: Z = (x_bar - mean) / std_err
z_stat = (x_bar - Ho) / std_err
p_val = 2*norm.sf(abs(z_stat))
print('z-stat: ', z_stat)
print('p-val: ', p_val)
```

$\alpha=.10,.05,.01$

1. the p val is less than $\alpha=.10$, so we reject the Ho
2. the pval is less than $\alpha=.05$, so we reject the Ho
3. the pval is greater than $\alpha=.01$, so we fail to reject the Ho

```python
CI_95_LB = x_bar - (1.960 * std_err)
CI_95_UB = x_bar + (1.960 * std_err)
print(f'The CI for mu: ({CI_95_LB}, {CI_95_UB})')
```

Question 2
---
Suppose
   $$
   \hat{S}=3.1,\qquad \widehat{SE}(\hat{S})=1.3.
   $$
   Construct 90%, 95%, and 99% confidence intervals, and test the null hypothesis that $S=0$ at the corresponding significance levels $\alpha=.10,.05,.01$. When do you reject or fail to reject the null of $S=0$?

```python
s_hat = 3.1
std_err = 1.3

CI_90_LB = s_hat - (1.645 * std_err)
CI_90_UB = s_hat + (1.645 * std_err)

CI_95_LB = s_hat - (1.960 * std_err)
CI_95_UB = s_hat + (1.960 * std_err)

CI_99_LB = s_hat - (2.576 * std_err)
CI_99_UB = s_hat + (2.576 * std_err)

print(f'90% CI: ({CI_90_LB}, {CI_90_UB})')
print(f'95% CI: ({CI_95_LB}, {CI_95_UB})')
print(f'99% CI: ({CI_99_LB}, {CI_99_UB})')
```

1. When alpha = .10, we reject the null.
2. When alpha = .05, we reject the null.
3. When alpha = .01, we fail to reject the null.

We reject the null of S=0 when the confidence interval does not include 0.  
We fail to reject the null of S=0 when the confidence interval does include 0.

Question 3

A health survey compares a biomarker to a benchmark value of 12. The sample has
   $$
   \bar{X}=11.5,\qquad s=3,\qquad n=36.
   $$
   Is 12 a plausible value for the population mean? Construct a 95% confidence interval for $H_0:\mu=12$ versus $H_1:\mu\ne12$. Do you reject or fail to reject the null?

```python
x_bar = 11.5
s = 3
n = 36
std_err = s / np.sqrt(n)

CI_95_LB = x_bar - (1.960 * std_err)
CI_95_UB = x_bar + (1.960 * std_err)

print(F'95% CI: ({CI_95_LB}, {CI_95_UB})')
```

We fail to reject the null because 12 is in the 95% confidence interval for the mean. It is plausible for the population mean to equal 12. 

Question 4
---
A real estate analyst compares the average price of a house in two neighborhoods.
    Group A:
    $$
    \bar{X}_A=25,\qquad s_A=8,\qquad n_A=100
    $$
    Group B:
    $$
    \bar{X}_B=22,\qquad s_B=7,\qquad n_B=100.
    $$
    Compute the two-sample $Z$-statistic for testing equality of means. Construct an approximate 95% confidence interval for $\mu_A-\mu_B$.

```python
# Ho: mu_a - mu_b = 0, H1: mu_a - mu_b != 0
# SE = sqrt((s_a^2/n) + (s_b^2/n))
# z_stat = x_bar_a - x_bar_b - 0 / std err
# CI: x_bar_a - x_bar_b +- .95 # * std err

x_bar_a = 25
s_a = 8
n_a = 100

x_bar_b = 22
s_b = 7
n_b = 100

std_err = np.sqrt((np.square(s_a) / n_a) + (np.square(s_b) / n_b))
z_stat = (x_bar_a - x_bar_b - 0) / std_err
print(f'z-stat: {z_stat}')


CI_95_LB = x_bar_a - x_bar_b - (1.960 * std_err)
CI_95_UB = x_bar_a - x_bar_b + (1.960 * std_err)
print(f'CI: ({CI_95_LB}, {CI_95_UB})')
```

Because 0 is not in the 95% CI, we reject the null hypothesis and it is not plausible that there is no difference between the 2 population means.

    Compute the standard error of the difference in means and the corresponding test statistic for
    $$
    H_0:\mu_A-\mu_B=0.
    $$

Question 6

A product-quality team compares defect scores across two factories.
    Factory A:
    $$
    \bar{X}_A=3.2,\qquad s_A=1.5,\qquad n_A=49
    $$
    Factory B:
    $$
    \bar{X}_B=2.8,\qquad s_B=1.4,\qquad n_B=64.
    $$
    Is the difference statistically large relative to the noise? Compute the test statistic and confidence interval for $H_0:\mu_A-\mu_B=0$, and interpret the result at $\alpha=.05$.

```python
x_bar_a = 3.2
s_a = 1.5
n_a = 49

x_bar_b= 2.8
s_b= 1.4
n_b = 64

std_err = np.sqrt((np.square(s_a)/n) + (np.square(s_b)/n))

z_stat = (x_bar_a - x_bar_b - 0) / std_err
print(f'z-stat: {z_stat}')

CI_95_LB = x_bar_a - x_bar_b - (1.960 * std_err)
CI_95_UB = x_bar_a - x_bar_b + (1.960 * std_err)
print(f'CI: ({CI_95_LB}, {CI_95_UB})')

p_val = 2*norm.sf(abs(z_stat))
print(f'pval: {p_val}')

```

The p val is 0.2421263762300413 which is greater than the alpha value, .05. At this alpha, we would fail to reject the null hypothesis because there is statistically significant evidence that the difference between the 2 means could be zero.

   b. Compute the bootstrap standard error as the sample standard deviation of the bootstrap estimates (using `ddof=1` ensures that there's the $1/(N-1)$ degrees of freedom correction).
   
   c. Using the original estimate and the bootstrap standard error, construct centered-Normal 90%, 95%, and 99% confidence intervals.

```python
data = pd.read_csv('./data/ames_prices.csv')
data
```

```python
a = np.arcsinh(data['area'])
p = np.arcsinh(data['price'])

c = a.corr(p)
print(f'correlation without bootstrapping: {c}')
```

```python
# a. use seed and bootstrap=2000
rng = np.random.default_rng(seed=123)
R = 2000
correlations = []
R_count = 0

for r in range(R):
    sample = data.sample(
        n=len(data),
        replace=True,
        random_state=rng
    )

    area = np.arcsinh(sample['area'])
    price = np.arcsinh(sample['price'])

    corr = area.corr(price)

    correlations.append(corr)
    R_count +=1

print(f'Seed: 123')
print(f'Bootstrap replications: {R_count}')
  
```

```python
# Compute the bootstrap standard error as the sample standard deviation of 
# the bootstrap estimates (using `ddof=1` ensures that there's the $1/(N-1)$ 
# degrees of freedom correction).

standard_error = np.std(correlations, ddof=1)
print(f'standard error of bootstrap estimates: {standard_error}')

   
```

```python
#  c. Using the original estimate and the bootstrap standard error, 
# construct centered-Normal 90%, 95%, and 99% confidence intervals.

# 90%:
CI_LB_90 = c - (1.645 * standard_error)
CI_UB_90 = c + (1.645 * standard_error)

#95:
CI_LB_95 = c - (1.960 * standard_error)
CI_UB_95 = c + (1.960 * standard_error)

#99
CI_LB_99 = c - (2.576 * standard_error)
CI_UB_99 = c + (2.576 * standard_error)

print(f'90% CI: ({CI_LB_90},{CI_UB_90})')
print(f'95% CI: ({CI_LB_95},{CI_UB_95})')
print(f'99% CI: ({CI_LB_99},{CI_UB_99})')
```
